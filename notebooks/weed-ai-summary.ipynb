# Weed AI Dataset Summary

The purpose of this notebook is to summarise the metadata extracted from the dataset pages from the Weed-AI repository.

```python
import pandas as pd

df = pd.read_json("./../all_datasets.json")
df.head()
```

```python
dataset_summary = df[["name", "published", "images"]].sort_values("images", ascending=False)
```

```python
def dataframe_to_typst(df: pd.DataFrame) -> str:
    lines = [
        "#table(",
        f"  columns: {len(df.columns)},"
    ]

    # Header row
    headers = ", ".join(f"[*{col}*]" for col in df.columns)
    lines.append(f"  {headers},")

    # Data rows
    for row in df.itertuples(index=False):
        row_text = ", ".join(f"[{value}]" for value in row)
        lines.append(f"  {row_text},")

    lines.append(")")

    return "\n".join(lines)
```

```python
print(dataframe_to_typst(dataset_summary))
```

```python
# Extract the weeds that we have available:
weed_species = pd.json_normalize(
    df["annotation_statistics"].explode().dropna()
).reset_index(drop=True).sort_values("category")

exploded = df.explode("annotation_statistics")
result = pd.concat(
    [
        pd.json_normalize(exploded["annotation_statistics"]).reset_index(drop=True),
        exploded["name"].reset_index(drop=True),
    ],
    axis=1
)
```

```python
print(dataframe_to_typst(result))
```
